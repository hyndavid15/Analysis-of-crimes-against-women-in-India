```python
df = df.dropna()
df
```

```python
## Remove duplicates
df.drop_duplicates(inplace = True)
uni_rows = df.shape[0]
uni_rows
```

```python
# To View Unique data
df.reset_index()
```

```python
null_value = df.isnull() == True
df.fillna(np.nan, inplace = True)
df                                   # we replace all the null value as NaN.
```

```python
df['State'] = df['State'].str.title()
df
```

## Data Sorting

```python
# Top 5 years with highest DV
top_dv_years = df.sort_values(by='DV', ascending=False).head(5)
top_dv_years
```

```python
#   Sort by State and Year
sorted_df = df.sort_values(by=['State', 'Year'])
sorted_df
```

```python
##Sort by Year (oldest to newest)
df.sort_values(by='Year', ascending=True)
```

## Data Filtering

```python
#Rape cases > 1000
high_rape = df[df['Rape'] > 1000]
high_rape
```

```python
#  Filter year = 2015 and DV > 2000
df_2015_dv = df[(df['Year'] == 2015) & (df['DV'] > 2000)]
df_2015_dv
```

```python
#  Filter data for specific states
south_states = df[df['State'].isin(['Andhra Pradesh', 'Tamil Nadu', 'Karnataka'])]
south_states
```

## Grouping and  Aggregation

```python
rape_by_state = df.groupby('State')['Rape'].sum().reset_index()
rape_by_state
```

```python
## Mean DV and AoW by year
avg_by_year = df.groupby('Year')[['DV', 'AoW']].mean().reset_index()
avg_by_year
```

```python
# Max rape cases per year
max_rape_year = df.groupby('Year')['Rape'].max()
max_rape_year
```

```python
# Min AoM by state
min_aom_state = df.groupby('State')['AoM'].min()
min_aom_state
```

```python
df[['Rape', 'DV']].agg(['min', 'max', 'mean', 'var'])
```

```python
#  Total crimes sum (across selected columns)
np.sum(df[['Rape', 'DV', 'AoW']].values)
```

## Data visualization

```python
# Line Plot
plt.plot(df['Year'], df['Rape'])
plt.title("Rape Cases Over Years")
plt.xlabel("Year")
plt.ylabel("Rape Cases")
plt.show()
```

```python
# Bar Chart
states = df['State'].head(10)
rape_cases = df['Rape'].head(10)

plt.bar(states, rape_cases)
plt.xticks(rotation=45)
plt.title("Rape Cases by State (Top 10)")
plt.ylabel("Cases")
plt.tight_layout()
plt.show()
```

```python
## Scatter Plot
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Rape', y='DV', hue='Year', palette='viridis', alpha=0.7)
plt.title("Rape vs Domestic Violence (Color by Year)", fontsize=14)
plt.xlabel("Rape Cases")
plt.ylabel("Domestic Violence Cases")
plt.legend(title='Year', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Histogram
plt.hist(df['DV'], bins=20)
plt.title("Distribution of Domestic Violence Cases")
plt.xlabel("DV Cases")
plt.ylabel("Frequency")
plt.show()
```

```python
## Box Plot
sns.boxplot(x='Year', y='DV', data=df)
plt.xticks(rotation=45)
plt.title("Domestic Violence Case Distribution per Year")
plt.show()
```

```python
## Box Plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Year', y='AoW', hue='Year', palette='Set2', legend=False)
plt.title("Distribution of Assault on Women (AoW) Cases per Year", fontsize=14)
plt.xlabel("Year")
plt.ylabel("AoW Cases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Pie Chart
#Group by State and get top 5 by total Rape cases
top5 = df.groupby('State')['Rape'].sum().sort_values(ascending=False).head(5)

# Plot pie chart
plt.figure(figsize=(6, 6))
plt.pie(top5, labels=top5.index, autopct='%1.1f%%', startangle=140)
plt.title("Top 5 States by Total Rape Cases (Pie Chart)")
plt.axis('equal')  # Make the pie chart a circle
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

#  Normalize numerical features using MinMaxScaler
scaler = MinMaxScaler()
scaled_df = scaler.fit_transform(df[['Rape', 'DV', 'AoW']])
scaled_df

```

```python
##Standardize data using StandardScaler
from sklearn.preprocessing import StandardScaler
std_scaler = StandardScaler()
standardized = std_scaler.fit_transform(df[['Rape', 'DV', 'AoW']])
standardized
```

```python
#  Convert to DataFrame for inspection
standardized_df = pd.DataFrame(standardized, columns=['Rape_scaled', 'DV_scaled', 'AoW_scaled'])
standardized_df
```

```python
# Heatmaps
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='Spectral', linewidths=0.5)
plt.title("Correlation Between Crime Metrics", fontsize=14)
plt.show()
```

```python
df['Total_Crimes'] = df[['Rape', 'DV', 'AoW', 'AoM', 'K&A', 'WT']].sum(axis=1)

#  Group by State and sum total crimes
total_crimes_by_state = df.groupby('State')['Total_Crimes'].sum().sort_values(ascending=False)

#  Select top 6 states
top6_crimes = total_crimes_by_state.head(6)

#  Plot pie chart
plt.figure(figsize=(7, 7))
plt.pie(top6_crimes, labels=top6_crimes.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.title("Top 6 States by Total Crimes Against Women")
plt.axis('equal')  # Equal aspect ratio ensures pie is drawn as a circle
plt.show()
```
